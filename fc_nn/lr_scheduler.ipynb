```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.utils.data as data
```

```python
class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(10, 1)

    def forward(self, x):
        return self.linear(x)
```

```python
model = Model()
optimizer = optim.Adam(model.parameters(), lr=0.01)
loss_func = nn.MSELoss()
```

### Определяем scheduler

```python
scheduler = optim.lr_scheduler.ExponentialLR(optimizer=optimizer, gamma=0.9) # gamma - lr умножается на это число при step()
```

```python
data_x = torch.ones(100, 10)
data_y = torch.ones(100, 1)

ds = data.TensorDataset(data_x, data_y)
data_train = data.DataLoader(ds, batch_size=1, shuffle=True)
```

```python
for x_batch, y_batch in data_train:
    y_pred = model(x_batch)
    loss = loss_func(y_pred, y_batch)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    # делаем шаг для уменьшения lr
    scheduler.step()
```
